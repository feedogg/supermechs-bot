```python
import orjson
import requests

URL_V1 = "https://gist.githubusercontent.com/ctrlraul/3b5669e4246bc2d7dc669d484db89062/raw"
URL_V2 = "https://gist.githubusercontent.com/ctrlraul/22b71089a0dd7fef81e759dfb3dda67b/raw"
URL_V3 = "https://raw.githubusercontent.com/Enegg/Item-packs/master/items.json"

session = requests.Session()

with session.get(URL_V1) as response:
    response.raise_for_status()
    data_v1 = orjson.loads(response.content)

with session.get(URL_V2) as response:
    response.raise_for_status()
    data_v2 = orjson.loads(response.content)

with session.get(URL_V3) as response:
    response.raise_for_status()
    data_v3 = orjson.loads(response.content)

session.close()
```

```python
from parse.common import converter
from parse.generic import ItemPack
```

```python
from parse.v1 import structure

raw_pack_v1 = converter.structure(data_v1, ItemPack)
pack_v1 = structure(raw_pack_v1)
```

```python
from parse.v2 import structure

pack_v2 = structure(data_v2)
pack_v3 = structure(data_v3)
```

```python
from parse.v3 import structure

pack_v3 = structure(data_v3)
```
